```python
import csv
with open('emp.csv','w',newline='') as f:
  w=csv.writer(f)
  w.writerow(('END','NAME','SALARY','ADDR'))
  n=int(input('Enter no of enployees: '))
  for i in range(n):
    no= input('enter enpolyee no:')
    name= input('enter enpolyee name:')
    sal= input('enter enpolyee sal:')
    addr= input('enter enpolyee addr:')
    w.writerow([no,name,sal,addr])

print('data inserted successfully!')
```

```python
# reading data from csv file
import csv
f= open('emp.csv','r')
r= csv.reader(f)
data= list(r)
print(data)
for line in data:
  for word in line:
    print(word,'\t',end='')
  print()
```
