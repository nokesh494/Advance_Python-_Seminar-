```python
import numpy as np
students = np.array([
    [85, 56, 48],
    [78, 65, 89],
    [75, 32, 14],
    [36, 47, 58],
    [10, 20, 60]
])
students
```

```python
avg_marks = np.mean(students, axis=0)
print('Average marks: ',avg_marks)
```

```python
topper = np.argmax(avg_marks)
print('topper student: ',topper)
```

### Why is the topper output `2`?

The reason the final code output is `topper student: 2` is due to a variable name conflict in your cells:

1. **Step 1 (Student Averages):** In cell `WtkNw87mxNTj`, you calculated the average mark per *student* (row-wise, using `axis=1`) and saved it as `avg_marks`:
   * `avg_marks` = `[63.0, 77.33, 40.33, 47.0, 30.0]` (representing each of the 5 students).

2. **Step 2 (Subject Averages - Overwriting):** In cell `X0RIUAkByDG1`, you recalculated `avg_marks` but with `axis=0` (column-wise average, representing each of the 3 subjects):
   * This **overwrote** the previous value of `avg_marks` with: `[34.6, 46.6, 73.4]`.

3. **Step 3 (Finding the Topper):** In cell `jysZCgB3yVup`, you ran `np.argmax(avg_marks)`:
   * Because `avg_marks` was overwritten, it checked the 3-element subject averages `[34.6, 46.6, 73.4]`.
   * The highest value in this array is `73.4`, which is located at **index 2** (the third subject). Therefore, the output is `2` instead of the index of the highest-scoring student (who would have been index `1` with an average of `77.33`).

```python
#Bar Chart
months = ['Jan', 'Feb', 'Mar', 'April']

sales = [250, 300, 400, 350]

plt.bar(months, sales, color='skyblue')

plt.title('Monthly sales')

plt.xlabel('Month')

plt.ylabel('Sales')

plt.show()
```

```python
plt.barh(months, sales, color='orange')

plt.title('Monthly sales (Horizontal)')

plt.xlabel('sales')

plt.ylabel('months')

plt.show()
```

```python
#Scatter Plot(comparing Height vs Weight)
height = [150, 160, 170, 180]
weight= [50, 60, 65, 75]
plt.scatter(height, weight, color='green')
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.grid(True)
plt.show()
```

```python
#pie Chart(market Share)
brands=['Brand A', 'Brand B', 'Brand C']
shares=[45, 38, 25]
plt.pie(shares, labels=brands, autopct='%1.1f%%', startangle=True)
plt.title('Market Shares')
plt.axis('equal')
plt.show()
```

```python
#Multiple Plot in one figure
x = [1,2,3,4]
y1= [1,4,9,16]
y2= [1,2,3,4]
plt.plot(x, y1, label='Square')
plt.plot(x, y2, label='Linear')
plt.legend()
plt.title('Multiple Lines')
plt.grid(True)
plt.show()
```

```python
#Subplots
fig, axs = plt.subplots(2, 2)

# Top-Left: Square
axs[0, 0].plot(x, y1)
axs[0, 0].set_title('Square')

# Top-Right: Sales
axs[0, 1].bar(months, sales)
axs[0, 1].set_title('Sales')

# Bottom-Left: Scatter
axs[1, 0].scatter(height, weight)
axs[1, 0].set_title('Scatter')

# Bottom-Right: Pie Chart
axs[1, 1].pie(shares, labels=brands, autopct='%1.1f%%')
axs[1, 1].set_title('Pie')

plt.tight_layout()
plt.show()
```
