```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Name':['A', 'B', 'C', 'D', 'E'],
    'Math': [78, 67, 98, 56, 88],
    'Science':[85, 72, 95, 60, 84],
    'English': [88,70, 92, 58, 86]
}
df=pd.DataFrame(data)
print(df)
```

```python
marks = df[["Math", "Science", "English"]].values
average_marks = np.mean(marks,axis=1)
df["Average"] = average_marks
print(df)
```

```python
def grade(avg):
  if avg >=90:
    return "A"
  elif avg >= 75:
    return "B"
  elif avg >= 60:
    return "C"
  else:
    return "D"

df['Grade'] = df['Average'].apply(grade)
df
```

```python
plt.figure()
plt.bar(df["Name"], df["Average"])
plt.xlabel('Students')
plt.ylabel('Average Marks')
plt.title('Average Marks of Students')
plt.show()
```

```python
#Average marks per subject
df[["Math", "Science", "English"]].mean().plot(kind="bar")
plt.title("Average Marks per Subject")
plt.ylabel("Marks")
plt.show()
```
