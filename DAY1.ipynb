```python
class StudentAdmission:
    def __init__(self, name, roll_no, marks_list):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks_list
        self.total = sum(marks_list)
        self.percentage = (self.total / 500) * 100

    def check_admission(self):
        print(f"\n--- Student Details ---")
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")
        print(f"Total Marks: {self.total} / 500")
        print(f"Percentage: {self.percentage:.2f}%")

        if self.percentage >= 85:
            print("Status: Admission Approved! (Percentage is 85% or above)")
        else:
            print("Status: Admission Rejected (Percentage is below 85%)")

# Input collection
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")
marks = []
print("Enter marks for 5 subjects (out of 100 each):")
for i in range(1, 6):
    mark = float(input(f"Subject {i} Marks: "))
    marks.append(mark)

# Creating object and calling method
student = StudentAdmission(name, roll_no, marks)
student.check_admission()
```

```python
#single level inheritance
class a:
  def showa(self):
     print('showa funftion')

class b(a):
  def showb(self):
    print('showb function ')

obj=b()
obj.showa()
obj.showb()
```

```python
class a:
  def showa(self):
     print('showa funftion')

class b:
  def showb(self):
    print('showb function ')

class c(a,b):
  def showc(self):
    print('showc function ')

obj=c()
obj.showa()
obj.showb()
obj.showc()
```

```python
class d:
  def showd(self):
    print('showd function ')

class a(d):
  def showa(self):
     print('showa function')

class b(d):
  def showb(self):
    print('showb function ')

class c(d):
  def showc(self):
    print('showc function ')

# To access methods, we create instances of the subclasses.
# Subclasses can access their own methods AND the parent class (d) method.
obj_a = a()
obj_a.showa()
obj_a.showd()

obj_b = b()
obj_b.showb()
obj_b.showd()

obj_c = c()
obj_c.showc()
obj_c.showd()
```
